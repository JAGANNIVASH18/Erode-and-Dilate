```python
# Import the necessary packages
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img2 = np.zeros((100,400), dtype='uint8')

font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(
    img2,
    'TheAILearner',
    (5,70),
    font,
    2,
    (255),
    5,
    cv2.LINE_AA
)
```

```python
kernel = np.ones((5,5), np.uint8)

kernel1 = cv2.getStructuringElement(
    cv2.MORPH_CROSS,
    (7,7)
)
```

```python
image_erode1 = cv2.erode(img2, kernel1)
```

```python
image_dilate1 = cv2.dilate(img2, kernel1)
```

```python
titles = ["Original Image", "Eroded Image", "Dilated Image"]
images = [img2, image_erode1, image_dilate1]
for i in range(0,3):

    plt.figure(figsize=(10,10))

    plt.subplot(1,2,1)
    plt.title("Original Image")
    plt.imshow(img2, cmap='gray')
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.title(titles[i])
    plt.imshow(images[i], cmap='gray')
    plt.axis("off")

    plt.show()
```
